```python
def calcular_desconto(*valores , desconto: int = 0):
  total = sum(valores)
  total_com_desconto = total - (total * desconto / 100)
  return total_com_desconto

print(calcular_desconto(100,200, desconto = 10))
print(calcular_desconto(100,200))
```
