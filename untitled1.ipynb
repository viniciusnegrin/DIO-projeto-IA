Criando listas e modificando com for e if

```python
lista = [i for i in range (21)]
lista_pares = [i for i in lista if i % 2 == 0]
lista_impar = [i for i in lista if i % 2 != 0]
lista_quad = [i ** 2 for i in lista]
double_quad = [i * 2 for i in lista_quad]
lista_somapar = lista_pares + lista_quad

print(lista)
print(lista_pares)
print(lista_impar)
print(lista_quad)
print(double_quad)
print(sorted(lista_somapar))
print(lista+lista_impar)
```
